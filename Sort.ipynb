Sort() function
- can be written as np.sort() or arr.sort()

Parameters
- arr = array to sort 
- axis = columns or rows to sort
- kind = quicksort, mergesort, heapsort
- order = on which field we want to sort

np.dtype() function
- allorts datatypes to a specific parameter in an array

```python
import numpy as np

a = np.array([2,4,6,8,1,3,-9])
print(f"before sort: {a}")
a.sort()
print(f"after sort: {a}")

dt = np.dtype([('name','S10'),('age',int)])
b = np.array([('Darshan',20),('ABC',30)('XYZ',10)], dtype = dt)
b.sort(order = 'name')
print(b)
```
