```python
import numpy as np

l = [i for i in range(1,26)]
print(l)
a = np.array(l)
print(a)
reshape = a.reshape(5,5)
print(reshape)

print(reshape[2][3])
print(reshape[2,3])
print(reshape[2])
print(reshape[0:2])
print(reshape[0:2:2])
print(reshape[::-2])
print(reshape[1:3,1:3])
print(reshape[3::3])
print(reshape[::-1])
```
