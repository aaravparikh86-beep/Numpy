array function - to create an array
Syntax: np.array()

```python
import numpy as np

List = np.array([10,20,30,40,50])
print(List)

Tuple = np.array((10,30,50,70))
print(Tuple)

Set = np.array({1,1,1,1,3,4,5,6,8})
print(Set)

print(type(List))
print(type(Tuple))
print(type(Set))
```

arrange function - similar to range function
Syntax: arrange(start,end,step)

```python
import numpy as np

Arrange = np.arange(1,11,2)
print(Arrange)
```

Zeros function - prints zeros in the dimension mentioned
Syntax: np.zeros(dimension,row,column)

```python
import numpy as np

n = np.zeros((2,2,2))
print(n)
```

One function - prints ones in the dimension mentioned
Syntax: np.ones(dimension,row,column)

```python
import numpy as np

n = np.ones((2,2,2))
print(n)
```

eye function - make a square of number inputed and writes 1 from left corner to right corner in slanting way
Syntax: np.eye(side)

```python
import numpy as np

e = np.eye(6)
print(e)
```

shape function - to see the shape of the array
Syntax:(e.shape)

```python
import numpy as np

e = np.eye(6)
print(e)
print(e.shape)
```

linespace function - prints random number 
linespace (start, stop, nums of elements)

```python
import numpy as np

s = np.linspace(1,2,20)
print(s)
```

random.randint function - picks random number
Syntax: np.random.randint(starting number, ending number, number of numbers)
reshape function - reshapes the array
Syntax: re1.reshape(rows,columns)

```python
import numpy as np

re1 = np.random.randint(1,100,15)
print(re1)
re2 = re1.reshape(5,3)
print(re2)
```

ndim function = instead of making more dimension we can use ndim
Syntax: ndim = number of dimension

```python
import numpy as np

k = np.array([[1,2,3,4],[5,6,7,8]],ndmin=6)
print(k)
print(k.shape)
print(k.ndim)
```
